### Converting the Json files to a readeable dataset format

```python
import os
import json
import pandas as pd

INPUT_FOLDER = r"C:\Users\HP\Downloads\2021.naacl-main.472.OptionalSupplementaryData\QMSum\ALL\train"
OUTPUT_CSV = "dataset.csv"
OUTPUT_XLSX = "dataset.xlsx"

rows = []

def build_conversation(transcripts):
    convo = ""
    for turn in transcripts:
        speaker = turn.get("speaker", "")
        text = turn.get("content", "")
        convo += f"{speaker}: {text}\n"
    return convo

for filename in os.listdir(INPUT_FOLDER):
    if not filename.endswith(".json"):
        continue

    filepath = os.path.join(INPUT_FOLDER, filename)

    with open(filepath, "r", encoding="utf-8") as f:
        data = json.load(f)

    conversation = build_conversation(data["meeting_transcripts"])

    # ----- General queries -----
    for q in data.get("general_query_list", []):
        rows.append({
            "Conversation": conversation.strip(),
            "Query": q["query"],
            "Answer": q["answer"],
            "Source_File": filename
        })

    # ----- Specific queries -----
    for q in data.get("specific_query_list", []):
        span = q.get("relevant_text_span", [])

        if span:
            start = int(span[0][0])
            end = int(span[0][1])
            snippet = conversation[start:end]
        else:
            snippet = conversation

        rows.append({
            "Conversation": snippet.strip(),
            "Query": q["query"],
            "Answer": q["answer"],
            "Source_File": filename
        })

# Create dataframe
df = pd.DataFrame(rows)

# Save outputs
df.to_csv(OUTPUT_CSV, index=False)
df.to_excel(OUTPUT_XLSX, index=False)

print("Dataset successfully created!")
print(f"Rows: {len(df)}")
print(f"Files saved: {OUTPUT_CSV}, {OUTPUT_XLSX}")
```
